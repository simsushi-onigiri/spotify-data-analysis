```python
import pandas as pd
df = pd.read_csv('spotify dataset.csv')
```

```python
df['artists'] = df.groupby('album_name')['artists'].transform(
    lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'Unknown')
)
df['album_name'] = df.groupby('artists')['album_name'].transform(
    lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'Unknown')
)
df['track_name'] = df.groupby('popularity')['track_name'].transform(
    lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'Unknown')
)
```

```python
df[df['artists'].isnull() | df['album_name'].isnull()]
```

```python
df = df.dropna(subset=['artists', 'album_name'])
```

```python
df.isnull().sum()
```

```python
df = df.drop(['Unnamed: 0', 'key', 'mode'], axis=1)
```

```python
df = df.reset_index(drop=True)  # renumbers cleanly from 0
df.index = df.index + 1          # shifts to start from 1
```

```python
df.head()
```

```python
df.to_csv('spotify_data_cleaned.csv', index=False)
```
