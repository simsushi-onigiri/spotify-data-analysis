```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv('spotify_data_cleaned.csv')
```

```python
# 7.Top 10 vs Bottom 10 artists comparison based on 
# the audio features: danceability, energy, valence and tempo. 

features = ['popularity', 'danceability', 'energy', 'valence', 'tempo']

corr_matrix = df[features].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    annot=True,        # show the actual correlation numbers on the plot
    fmt='.2f',
    cmap='RdBu',
    vmin=-1, vmax=1,   # fixes the color scale so 0 is always neutral/white
    center=0
)
plt.title('Correlation: Popularity vs Audio Features')
plt.tight_layout()
plt.show()
```
